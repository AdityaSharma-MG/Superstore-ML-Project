## 1) Importing Data

```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('../data/Superstore_prepared.csv',encoding='Latin1')
df.head()
```

## 2) import necessary ML modules

```python
from sklearn.model_selection import train_test_split
# Models that will be trained
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor
```

## 3) Split the dataset in training and testing

```python
# Select the predictor set and target set
X=df.drop("Profit", axis=1)
Y=df["Profit"]
```

```python
Xtrain,Xtest,Ytrain,Ytest=train_test_split(X,Y,test_size=0.2,random_state=42)
```

## 4) Scaling of predictor columns

```python
from sklearn.preprocessing import StandardScaler

scaler=StandardScaler()
Xtrain_s=scaler.fit_transform(Xtrain)
XTest_s=scaler.transform(Xtest)
```

```python
# we will use the scaled data for Linear Regression
```

## 5) Linear regression

```python
lr=LinearRegression()
lr.fit(Xtrain_s,Ytrain)
ypred_lr=lr.predict(XTest_s)
```

## 6) Decision Tree Regression

```python
dt=DecisionTreeRegressor(random_state=42)
dt.fit(Xtrain,Ytrain)
ypred_dt=dt.predict(Xtest)
```

## 7) Random Forest

```python
rf=RandomForestRegressor(random_state=42)
rf.fit(Xtrain,Ytrain)
ypred_rf=rf.predict(Xtest)
```

## 8) Gradient Boosting

```python
gb=GradientBoostingRegressor(random_state=42)
gb.fit(Xtrain,Ytrain)
ypred_gb=gb.predict(Xtest)
```

## 9) Linear Regression with PCA

```python
from sklearn.decomposition import PCA

pca=PCA(n_components=0.9)

Xtrain_pca=pca.fit_transform(Xtrain_s)
Xtest_pca=pca.transform(XTest_s)
```

```python
lr2=LinearRegression()
lr2.fit(Xtrain_pca,Ytrain)
ypred_lr_pca=lr2.predict(Xtest_pca)
```

## 10) Model Evaluation

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score

#storing all results
results = []
```

```python
# For Linear regression

mae = mean_absolute_error(Ytest, ypred_lr)
mse = mean_squared_error(Ytest, ypred_lr)
rmse = np.sqrt(mse)
r2 = r2_score(Ytest, ypred_lr)
```

```python
results.append({
    "Model": "Linear Regression",
    "MAE": mae,
    "RMSE": rmse,
    "R²": r2,
})
```

```python
# For Decision Tree

mae = mean_absolute_error(Ytest, ypred_dt)
mse = mean_squared_error(Ytest, ypred_dt)
rmse = np.sqrt(mse)
r2 = r2_score(Ytest, ypred_dt)
```

```python
results.append({
    "Model": "Decision Tree Regressor",
    "MAE": mae,
    "RMSE": rmse,
    "R²": r2
})
```

```python
# For Random Forest Regressor

mae = mean_absolute_error(Ytest, ypred_rf)
mse = mean_squared_error(Ytest, ypred_rf)
rmse = np.sqrt(mse)
r2 = r2_score(Ytest, ypred_rf)
```

```python
results.append({
    "Model": "Random Forest Regressor",
    "MAE": mae,
    "RMSE": rmse,
    "R²": r2
})
```

```python
# For Gradient Boosting regressor

mae = mean_absolute_error(Ytest, ypred_gb)
mse = mean_squared_error(Ytest, ypred_gb)
rmse = np.sqrt(mse)
r2 = r2_score(Ytest, ypred_gb)
```

```python
results.append({
    "Model": "Gradient Boosting Regressor",
    "MAE": mae,
    "RMSE": rmse,
    "R²": r2
})
```

```python
# For Linear regression with PCA

mae = mean_absolute_error(Ytest, ypred_lr_pca)
mse = mean_squared_error(Ytest, ypred_lr_pca)
rmse = np.sqrt(mse)
r2 = r2_score(Ytest, ypred_lr_pca)
```

```python
results.append({
    "Model": "Linear Regression With PCA",
    "MAE": mae,
    "RMSE": rmse,
    "R²": r2
})
```

```python
# Final Comparison

results_df=pd.DataFrame(results)
results_df
```

```python
# we prefer highest Rsquared value and lowest RMSE
#so Gradient Boosting Regressor is the clear choice for us
```

## Conclusion

- Five regression models were evaluated.
- Gradient Boosting Regressor achieved the best performance (R² = 0.216, RMSE = 252.56).
- Random Forest was the second-best model.
- Linear Regression struggled, indicating the relationship between features and profit is non-linear.
- PCA improved Linear Regression but did not outperform ensemble tree methods.
