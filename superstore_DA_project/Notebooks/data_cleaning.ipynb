## 4)Change types of Shipment Date and Order Date

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python
df["Order Date"].head()
```

## 5)Creating new columns-

### delivery time , year of order , month of order , day of order

```python
# This column shows the number of days for delivery
df["Delivery Time"]=(df["Ship Date"]-df["Order Date"]).dt.days
```

```python
df["Delivery Time"].head()
```

```python
df["Year of order"]=df["Order Date"].dt.year
df["Month of order"]=df["Order Date"].dt.month
df["Weekday of order"]=df["Order Date"].dt.weekday
```

```python
# This column shows the year of order
df["Year of order"].head()
```

```python
# This column shows the month number of order
df["Month of order"].head()
```

```python
# This column shows the day of week(in number) of order
df["Weekday of order"].tail()
```

```python
df["Order Date"] = df["Order Date"].dt.strftime("%Y-%m-%d")
df["Ship Date"] = df["Ship Date"].dt.strftime("%Y-%m-%d")
```
