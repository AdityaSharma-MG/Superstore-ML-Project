## 3) Removing unwanted columns for model training

```python
# dont need unique /identifier attributes like-
dropped=[]
# Row id, Order ID, customer id, customer name, product id, product name
```

```python
arr1=['Row ID', 'Order ID','Customer ID', 'Customer Name','Product ID','Product Name']
dropped.extend(arr1)
df=df.drop(columns=arr1)
```

```python
df.head()
```

```python
df['Country'].unique()
```

```python
# no value in ML model, also we have state and city- so no need for postal code(zipcode)
```

```python
arr2=['Country','Postal Code']
dropped.extend(arr2)
df=df.drop(columns=arr2)
```

```python
# we can remove columns Order Date and Ship date
# we've already extracted the year, month, weekday and time to ship
```

```python
arr3=['Order Date','Ship Date']
dropped.extend(arr3)
df=df.drop(columns=arr3)
```

```python
df.columns
```

```python
dropped
```

## 4) Encode categorical columns

```python
df['Ship Mode'].unique()
```

```python
# segment has 3 categories
# ship mode has 4 categories
# state has 49 categories
# region has 4 categories-can be dropped as state already considered
# category has 3 categories
# sub-category has 17 categories
```

```python
encoded = ['Ship Mode', 'Segment', 'State', 'Category', 'Sub-Category']
df = pd.get_dummies(df, columns=encoded, drop_first=True)
```

```python
# removing city as state gives enough variations
# 500 unique values can cause overfit
```

```python
df=df.drop(columns=['City','Region'])
dropped.extend(['City','Region'])
```
